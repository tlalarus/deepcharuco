```python
# 📦 라이브러리 임포트
import os
import torch
from pytorch_lightning import Trainer
from configs import load_configuration
from models.net import lModel, dcModel
from data import CharucoDataset
from torch.utils.data import DataLoader

os.environ["CUDA_LAUNCH_BLOCKING"] = "1" # 비동기연산 -> 디버깅 
os.environ["TORCH_USE_CUDA_DSA"] = "1"

print('torch:', torch.__version__)
print('torch.version.cuda:', torch.version.cuda)
print('is_available:', torch.cuda.is_available())
print('device_count:', torch.cuda.device_count())

# if torch.cuda.device_count() > 0:
#     torch.cuda.set_device(0)
#     x = torch.rand(1, device="cuda:0")
#     print("tensor device:", x.device, "name:", torch.cuda.get_device_name(0))

# ✅ GPU 사용 가능 여부
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
# 📁 config.yaml 로드
CONFIG_PATH = "config.yaml"
config = load_configuration(CONFIG_PATH)

# 코너 ID 개수 (추론과 학습 모두 필요)
n_ids = (config.row_count - 1) * (config.col_count - 1)
#n_ids = int((config.row_count * config.col_count) / 2)
print(f"Number of Charuco corner IDs: {n_ids}")
```

```python
# 📚 학습 데이터셋 & DataLoader 정의
train_dataset = CharucoDataset(config, config.train_labels, config.train_images, visualize=False, validation=True)
train_loader = DataLoader(train_dataset, batch_size=config.bs_train, shuffle=True, num_workers=0)

# 🧪 검증 데이터셋
val_dataset = CharucoDataset(config, config.val_labels, config.val_images, visualize=False, validation=True)
val_loader = DataLoader(val_dataset, batch_size=config.bs_val, shuffle=False, num_workers=0, pin_memory=True) #prefetch_factor=10
```

```python
# 🧠 모델 생성
model = lModel(dcModel(n_ids=n_ids))

# 🏋️ Trainer 정의 및 학습 시작
trainer = Trainer(
    max_epochs=5,
    accelerator="gpu" if torch.cuda.is_available() else "cpu",
    devices=1,
    limit_val_batches=0,
    detect_anomaly=True,
    precision="bf16",
    num_sanity_val_steps=0
)

trainer.fit(model, train_loader, val_loader)
```

```python
len(val_loader)
next(iter(val_loader))
```

```python
trainer.validate(model, dataloaders=val_loader)
```
